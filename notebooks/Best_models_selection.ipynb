```python
import pandas as pd
from pathlib import Path
```

```python
company = ['Apple', 'Exxon', 'JPM', 'PFE']
for i in company:
    folder = Path(f"result_table_{i}")

    csv_files = folder.glob("*.csv")

    # Read each file and store the DataFrames in a list
    frames = [pd.read_csv(f) for f in csv_files]

    combined = pd.concat(frames, ignore_index=True)

    top5 = combined.sort_values("AIC", ascending=True).head(5)

    top5.to_csv(folder / "top5.csv", index=False)
```
